```python
import pandas as pd

data = {
    'Month':['Jan','Feb','Mar','Apr','May','Jun'],
    'Sales':[200,250,300,280,350,400]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

plt.plot(df['Month'], df['Sales'])

plt.title("Sales Graph")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

df['Month_Number']=[1,2,3,4,5,6]

X=df[['Month_Number']]
y=df['Sales']

model=LinearRegression()

model.fit(X,y)

future_data = pd.DataFrame({'Month_Number':[7]})

prediction = model.predict(future_data)

print("Next Month Sales =",prediction)
```

```python
future_data = pd.DataFrame({'Month_Number':[7]})

prediction = model.predict(future_data)

plt.plot(df['Month_Number'],df['Sales'])

plt.scatter(7,prediction[0])

plt.xlabel("Month Number")
plt.ylabel("Sales")
plt.title("Sales Forecast Prediction")

plt.show()
```
